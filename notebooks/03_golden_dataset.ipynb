```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")
GOLDEN_DIR = Path("../data/golden")

GOLDEN_DIR.mkdir(parents=True, exist_ok=True)

files = sorted(DATA_DIR.glob("*.csv"))

for file in files:
    if file.stem == "data_dictionary":
        continue

    df = pd.read_csv(file)

    output = GOLDEN_DIR / f"{file.stem}_golden.csv"
    df.to_csv(output, index=False)

    print(f"Created: {output.name}")
```

```python
success_statuses = {
    "SUCCESS",
    "SUCCESSFUL",
    "COMPLETED",
    "PAID"
}

payments_clean["is_successful"] = (
    payments_clean["payment_status"]
    .astype(str)
    .str.upper()
    .isin(success_statuses)
)
```

```python
def save_golden(df, name):
    path = GOLDEN_DIR / f"{name}_golden.csv"
    df.to_csv(path, index=False)
    print(f"Saved {path}")
```

```python
save_golden(borrowers_clean, "borrowers")
save_golden(accounts_clean, "accounts")
save_golden(agents_clean, "agents")
save_golden(calls_clean, "calls")
save_golden(attempts_clean, "call_attempts")
save_golden(payments_clean, "payments")
```

```python
cleaning_impact = []
```

```python
tables = [
    ("borrowers", borrowers, borrowers_clean),
    ("accounts", accounts, accounts_clean),
    ("agents", agents, agents_clean),
    ("calls", calls, calls_clean),
    ("call_attempts", attempts, attempts_clean),
    ("payments", payments, payments_clean)
]

for name, raw, clean in tables:

    cleaning_impact.append({
        "dataset": name,
        "raw_rows": len(raw),
        "golden_rows": len(clean),
        "removed_rows": len(raw) - len(clean),
        "removed_pct": round(
            (len(raw) - len(clean)) / len(raw) * 100,
            2
        )
    })

cleaning_impact = pd.DataFrame(cleaning_impact)

display(cleaning_impact)
```

```python
cleaning_impact.to_csv(
    REPORT_DIR / "cleaning_impact.csv",
    index=False
)
```
